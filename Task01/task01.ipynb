Исследование влияние шума, начальных параметров и learning rate на сходимость градиентного спуска при квадратичной регрессии.

f(x) = 1 + 2 * x + 3 * x**2 + шум

Сравниваются:
    - 4 уровня шума: 0, 0.1, 0.5, 1
    - 6 learning rate: 0.001, 0.01, 0.1, 0.5, 0.7, 0.9
    - 3 начальных позиции точек: "0", "N(0, 1)", "U(1000, 2000)"

Оптимизация выполняется методом SGD с использованием среднеквадратичной функции потерь MSE.

```python
import pandas as pd
import torch
from matplotlib import pyplot as plt
from torch import optim

torch.manual_seed(42)
```

```python
def fparameters(a, b, c):
    return f"a={a.item():.6f}, b={b.item():.6f}, c={c.item():.6f}"
```

```python
def parameters(case):
    match case:
        case "0":
            return torch.zeros(1, requires_grad=True), torch.zeros(1, requires_grad=True), torch.zeros(1, requires_grad=True)
        case "N(0, 1)":
            return torch.randn(1, requires_grad=True), torch.randn(1, requires_grad=True), torch.randn(1, requires_grad=True)
        case "U(1000, 2000)":
            return torch.randint(1000, 2000, (1,), dtype=torch.float32, requires_grad=True), torch.randint(1000, 2000, (1,), dtype=torch.float32, requires_grad=True), torch.randint(1000, 2000, (1,), dtype=torch.float32, requires_grad=True)
```

```python
learning_rates = [0.001, 0.01, 0.1, 0.5, 0.7, 0.9]
noises = [0.0, 0.1, 0.5, 1.0]
parameter_cases = ["0", "N(0, 1)", "U(1000, 2000)"]

results = []

def generate_data(n_samples=100, noise_std=0.1):
    x = torch.linspace(-1, 1, n_samples)
    y = 1 + 2 * x + 3 * x**2 + noise_std * torch.randn(n_samples)
    return x, y


# Shuffles the indices
# Индексы перемешиваются без пересечений, чтобы создать случайный порядок данных.
# Это важно для обучения модели, чтобы избежать смещения и
# обеспечить разнообразие в обучающих и проверочных данных.
idx = torch.randperm(100)
# Train-valid split
train_idx = idx[:80]
val_idx = idx[80:]


def run(a, b, c, x_train, y_train, lr):
    n_epochs = 1000

    loss_history = []
    # Defines a SGD optimizer to update the parameters
    optimizer = optim.SGD([a, b, c], lr=lr)
    for _ in range(n_epochs):
        yhat = a + b * x_train + c * x_train**2
        error = y_train - yhat
        loss = (error**2).mean()
        loss_history.append(loss.item())
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()

    return loss_history


for noise_std in noises:
    x, y = generate_data(noise_std=noise_std)
    x_train, y_train = x[train_idx], y[train_idx]
    x_val, y_val = x[val_idx], y[val_idx]
    plt.figure("train")
    plt.title(f"train noise={noise_std}")
    plt.plot(x_train.numpy(), y_train.numpy(), "b", lw=0, marker="^")
    plt.show()

    plt.figure("result")
    for parameter_case in parameter_cases:
        init_params = parameters(parameter_case)
        for lr in learning_rates:
            a, b, c = [p.clone().detach().requires_grad_(True) for p in init_params]
            print(f"Initial values: {fparameters(a, b, c)}")
            loss_history = run(a, b, c, x_train, y_train, lr)
            print(f"Learning rate: {lr}")
            print(f"optimized values: {fparameters(a, b, c)}")
            with torch.no_grad():
                yhat = a + b * x_train
                yres = a + b * x_val
            val_error = torch.mean((y_val - yres) ** 2).item()
            train_error = torch.mean((y_train - yhat) ** 2).item()
            print(f"train error: {train_error:.6f}")
            print(f"validation error: {val_error:.6f}")
            print("=" * 50)
            plt.plot(loss_history, label=f"{parameter_case} lr={lr}")
            results.append({
                "noise": noise_std,
                "init": parameter_case,
                "lr": lr,
                "a": a.item(),
                "b": b.item(),
                "c": c.item(),
                "train_mse": train_error,
                "val_mse": val_error,
            })

        plt.title(f"noise={noise_std} a,b,c={parameter_case} 1000 epochs")
        plt.xlabel("Эпоха")
        plt.ylabel("MSE")
        plt.yscale("log")
        plt.legend()
        plt.show()
```

```python
df = pd.DataFrame(results)

display(df)
```

По полученным результатам можно отметить, что оптимальным learning rate является 0.7 при котором получаем устойчивое решение за меньшее количество итераций. Также можно увидеть, что при больших начальных параметрах решение сходится дольше.

Learning rate равный 0.9 постоянно увеличивает ошибку это связанно с увеличением шага скачка, что приводит к перескоку локального мимнимума.
